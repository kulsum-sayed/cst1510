## 2. Loading it from a file

Topic 3's `load_records` — the line-reading and splitting you already wrote, now
building the list up one dict at a time.

````python
def load_records(path):
    """Read label,value,limit lines from a file and return a list of dicts."""
    records = []
    with open(path) as f:
        for line in f:
            label, value, limit = line.strip().split(",")
            records.append({"label": label, "value": float(value), "limit": float(limit)})
    return records
````

## Topic 5 Challenge

Write `load_records(path)`, load records from a file of your own, process each one
(`OVER LIMIT` / `WARNING` / `OK`), and print a summary line: `"N of M were over
limit"`. Wrap the load in a `try`/`except` so a missing file prints a clear
message instead of crashing — Topic 4's problem, applied here.

````python
# your code here
def load_records(path):#making function load record with path as parameter to  see the path of file
    records=[] #empty list to store the dic in it later stepp
    try:
        with open(path,"r") as f: #open th epath and read the file
            for line in f: #for loop to read lines in th efile 
                label, value, limit = line.strip().split(",")#strip for space and spilt is for spliting the data with ,
                records.append({ # adding it to the empty records made above
                    "label":label,
                    "value":float(value), #getting it in float no string
                    "limit":float(limit)
                })
    except FileNotFoundError: #if file not found error is there the below msg would be printed
        print(f"File '{path}' does not exist") #using f string to get the error 
    return records
with open("sample.txt", "w") as f: #makig a file to test it
    f.write("R1,80,10\n") #writting the lines and \n for ending line
    f.write("R2,110,10\n")
    f.write("R3,2,10\n")

records=load_records("sample.txt") #loading it in the above records for output to se error line

overlimit_count = 0 # giving value zero so that when we loop we can increase the counter
total= 0 
for r in records:
    total += 1  # Add 1 to our count for every record we loop through
    percent= (r["value"]/r["limit"])*100 #calculating percenatge of value 
    if percent >= 100:
        status = "OVER LIMIT"#condition for overlimit
        overlimit_count += 1 #if it overlimit the count wouldbecome 1 from 0 and sso on
    elif percent >= 90:
        status="WARNING" 
    else:
        status="OK"
        
    print(f"{r['label']}:{status}") #prints the status

print(f"{overlimit_count} of {total} were over limit")
````

## Common mistakes

| Mistake | What you see | Fix |
|---|---|---|
| `records["label"]` instead of `records[0]["label"]` | `TypeError` — `records` is a list, not a dict | index into the list first, then key into the dict |
| building the dict with the wrong keys each time | `KeyError` later, inconsistently | keep every record's dict the same shape |
| forgetting `load_records` needs to `return` the list | `None` where you expected records | check every function actually returns what it builds |
